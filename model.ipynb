# 1. Top Inshights

Top 10 customers by sales


```python
top_10_customers = df.groupby('Customer Name')['Sales'].sum().sort_values(ascending=False).head(10)
top_10_customers
```

top product by sales


```python
top_products = df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(1)
(top_products)
```

sort subcategory by sales From largest to least


```python
sort_subcategory = df.groupby('Sub-Category')['Sales'].sum().sort_values(ascending=False)
sort_subcategory
```

# 2. Geographical Analysis

Sales by state

```python

sales_by_state = df.groupby('State')['Sales'].sum().sort_values(ascending=False)
sales_by_state
```

Sales by region

```python
sales_by_region = df.groupby('Region')['Sales'].sum().sort_values(ascending=False)
sales_by_region
```

# 3.Shipping Performance Analysis


 Average Delivery Duration


```python
#change datatype
df['Order Date'] = pd.to_datetime(df['Order Date'], dayfirst=True)

df['Ship Date'] = pd.to_datetime(df['Ship Date'], dayfirst=True)

#Calculate delivery duration
df['Delivery Duration'] = df['Ship Date'] - df['Order Date']

#Average
Average_delivery_duration = df['Delivery Duration'].mean()
Average_delivery_duration
```

Shipping Mode Performance

```python
shipping_stats = df.groupby('Ship Mode').agg({
    'Delivery Duration': 'mean',
    'Order ID': 'count',
    'Sales': 'sum'
}).rename(columns={
    'Delivery Duration': 'AVG Delivery Duration',
    'Order ID': 'Total Orders',
    'Sales': 'Total Sales'
}).sort_values(by='Total Sales', ascending=False)

shipping_stats
```
